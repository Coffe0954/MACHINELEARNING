9. Как отсортировать массив по n-му столбцу?

````python
import numpy as np

z = np.random.randint(0, 10, (4, 3))
print("Изначальный массив:\n", z)

n = 1
sorted_z = z[z[:, n].argsort()]
print(f"отсортированный по {n}-му столбцу массив:\n", sorted_z)
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
limit = df['BMI'].quantile(0.75)

high_bmi = df[df['BMI'] > limit]
normal_bmi = df[df['BMI'].between(30, 60)]

print('Высокий BMI:', len(high_bmi[high_bmi['Chronic Diseases'].notna()]) / len(high_bmi) * 100, '%')

print('BMI от 30 до 60:', len(normal_bmi[normal_bmi['Chronic Diseases'].notna()]) / len(normal_bmi) * 100, '%')
````
